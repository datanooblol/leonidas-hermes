```python
chunks = [
    "สวัสดีครับไม่แน่ใจ",
    "ไม่แน่ใจว่าสนใจติดต่อ",
    "เรื่องอะไรเป็นพิเศษไหม",
    "ครับหรืออยากสอบถามเกี่ยวกับ",
    "เกี่ยวกับสินค้าหรือโปรโมชั่นของเรา",
    "",
    "",
    "",
    "",
    "ได้ครับสักครู่นะครับ",
    "",
    "",
    "เรียบร้อยครับ ขอบคุณที่ใช้บริการ"
]
```

```python
def consolidate_transcriptions(chunks: list[str]):
    init_text = chunks[0]
    for chunk in chunks[1:]:
        init_text += deduplicate_exact_match(chunk, init_text)
    return init_text

consolidate_transcriptions(chunks)
    
```

```python
import requests
from typing import Optional
class LocalLLM:
    def __init__(self, model_name):
        self.base_url = "http://localhost:11434/api/chat"
        self.model_name = model_name

    def run(self, messages:list, system_prompt:Optional[str]=None):
        if system_prompt is None:
            system_prompt = "You are a helpful assistant."
        response = requests.post(
            self.base_url,
            json={
                "model": self.model_name,
                "messages": [
                    {"role": "system", "content": system_prompt},
                    *messages,
                ],
                "stream": False,
            },
            # timeout=600,
        )
        return response
model_name = "llama3.2-vision:11b"
# model_name = "llama3.2:1b"
llm = LocalLLM(model_name)

system_prompt = "Classify user's input into: positive, neutral, negative and return only result"
content = "I hate this pizza!"
messages = [
    {"role": "user", "content": content}
]
response = llm.run(messages=messages, system_prompt=system_prompt)
print(response.json()["message"]["content"])
```
