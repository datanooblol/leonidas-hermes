```python
from pathlib import Path
from pydub import AudioSegment

out_raw = Path("./out_raw")
out_wav = Path("./out_wav")
out_wav.mkdir(exist_ok=True, parents=True)

for file in out_raw.glob("*.webm"):
    try:
        print(file)
        audio = AudioSegment.from_file(file)
        print(f"Duration: {len(audio)}ms")
        print(f"Sample rate: {audio.frame_rate}Hz")
        
        # Save as WAV in out_wav directory with same name
        wav_file = out_wav / file.with_suffix('.wav').name
        audio.export(wav_file, format="wav")
        print(f"Saved as: {wav_file}")
    except Exception as e:
        print(f"Failed to process {file}: {e}")
```
