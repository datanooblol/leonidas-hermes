```python
from pydub import AudioSegment
from pathlib import Path

def merge_all_byte_files(input_dir, output_path):
    bin_files = sorted(Path(input_dir).glob("*.bin"))
    
    # รวมไฟล์ binary
    with open(output_path, 'wb') as output_file:
        for bin_file in bin_files:
            with open(bin_file, 'rb') as input_file:
                output_file.write(input_file.read())
    
    print(f"✅ Merged {len(bin_files)} files to {output_path}")
    return output_path

# ใช้งาน
merged_file = merge_all_byte_files("audio/recording_byte", "audio/merged_all.bin")
```

```python
import wave
import numpy as np

def merge_to_wav(input_dir, output_path, sample_rate=16000):
    bin_files = sorted(Path(input_dir).glob("*.bin"))
    all_data = b''
    
    for bin_file in bin_files:
        all_data += bin_file.read_bytes()
    
    # ตัดให้เป็นจำนวนคู่
    all_data = all_data[:len(all_data) - (len(all_data) % 2)]
    audio_data = np.frombuffer(all_data, dtype=np.int16)
    
    with wave.open(output_path, 'wb') as wav_file:
        wav_file.setnchannels(1)
        wav_file.setsampwidth(2)
        wav_file.setframerate(sample_rate)
        wav_file.writeframes(audio_data.tobytes())
    
    return output_path

merge_to_wav("audio/recording_byte", "audio/merged_all.wav")
```
