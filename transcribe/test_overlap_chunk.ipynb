step input
- file 1.wav, 2.wav

output
- 1.wav + x second from start of 2.wav

```python
from pydub import AudioSegment

def merge_audio_with_half_second(file1_path, file2_path, output_path, overlap:int=0.5):
    # โหลดไฟล์เสียงทั้งสองอัน
    audio1 = AudioSegment.from_wav(file1_path)
    audio2 = AudioSegment.from_wav(file2_path)
    
    # ตัด 0.5 วินาทีแรกจากไฟล์ที่สอง (500 มิลลิวินาที)
    overlap*=1000  # แปลงเป็นมิลลิวินาที
    audio2_half_second = audio2[:overlap]
    
    # รวมไฟล์แรกทั้งหมด + 0.5 วินาทีแรกของไฟล์ที่สอง
    merged_audio = audio1 + audio2_half_second
    
    # บันทึกไฟล์ผลลัพธ์
    merged_audio.export(output_path, format="wav")
    
    print(f"✅ Merged {file1_path} + 0.5s of {file2_path} to {output_path}")
    return output_path

# ใช้งาน
merge_audio_with_half_second("audio/chunks_fah/chunk_000.wav", "audio/chunks_fah/chunk_001.wav", "audio/merged_with_half.wav", overlap=1)
```
